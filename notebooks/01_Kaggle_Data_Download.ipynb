# **Data Download**

**Objective:** download the cherry leaves dataset from Kaggle into `inputs/raw/`.

**Input:** a Kaggle API token. **Output:** the images, exactly as published, in `inputs/raw/cherry-leaves/`.

Nothing in `inputs/` goes to git: the dataset is downloaded again on any machine.

## 1. Credentials

The Kaggle client reads the token from `~/.kaggle/kaggle.json` (or from the `KAGGLE_USERNAME` and `KAGGLE_KEY` environment variables). The token never lives inside the repository: create it at kaggle.com → Settings → API → *Create New Token*.

```python
from pathlib import Path

# The notebooks live in notebooks/; every path is relative to the repository root
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW = ROOT / "inputs" / "raw" / "cherry-leaves"
PREPARED = ROOT / "inputs" / "prepared"
OUTPUTS = ROOT / "outputs"
print("Repository root:", ROOT.name)
```

## 2. Download

The dataset is public: [ahmadhabibhasanzein/cherry-leaves](https://www.kaggle.com/datasets/ahmadhabibhasanzein/cherry-leaves).

```python
from kaggle.api.kaggle_api_extended import KaggleApi

DATASET = "ahmadhabibhasanzein/cherry-leaves"
if RAW.exists():
    print("Already downloaded:", RAW.relative_to(ROOT))
else:
    api = KaggleApi()
    api.authenticate()
    api.dataset_download_files(DATASET, path=RAW.parent, unzip=True, quiet=True)
    print("Downloaded to", RAW.relative_to(ROOT))
```
